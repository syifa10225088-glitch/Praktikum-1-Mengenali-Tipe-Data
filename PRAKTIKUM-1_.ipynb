```python
nama = "Fajar"
umur = 19
berat = 54.5
terdaftar = True

# 1. Tampilkan semua nilai variabel
print("Nama      :", nama)
print("Umur      :", umur)
print("Berat     :", berat)
print("Terdaftar :", terdaftar)

# 2. Tampilkan tipe data setiap variabel
print(type(nama))       # <class 'str'>
print(type(umur))       # <class 'int'>
print(type(berat))      # <class 'float'>
print(type(terdaftar))  # <class 'bool'>

# 3. Ubah umur menjadi 20
umur = 20
print("Umur      :", umur)
print(type(umur))
```
