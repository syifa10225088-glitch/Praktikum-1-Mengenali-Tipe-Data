```python
# ==========================================
# Tugas 1 & 2: Buat variabel dan periksa syarat
# ==========================================
print("--- DATA AWAL (Nilai: 82, Kehadiran: 90) ---")
nilai = 82
kehadiran = 90

# Memeriksa masing-masing syarat dengan operator >=
syarat_nilai = nilai >= 75
syarat_kehadiran = kehadiran >= 80

print(f"Apakah nilai memenuhi syarat (>= 75)? {syarat_nilai}")
print(f"Apakah kehadiran memenuhi syarat (>= 80)? {syarat_kehadiran}")

# ==========================================
# Tugas 3: Gunakan operator 'and'
# ==========================================
lulus_and = syarat_nilai and syarat_kehadiran
print(f"Status Kelulusan (menggunakan 'and'): {lulus_and}\n")

print("=" * 50 + "\n")

# ==========================================
# Tugas 4: Ubah nilai menjadi 70
# ==========================================
print("--- PERUBAHAN DATA (Nilai diubah menjadi 70) ---")
nilai = 70

# Periksa ulang syarat nilai
syarat_nilai = nilai >= 75
print(f"Apakah nilai baru memenuhi syarat (>= 75)? {syarat_nilai}")
print(f"Apakah kehadiran memenuhi syarat (>= 80)? {syarat_kehadiran}")

# Menggunakan 'and' kembali
lulus_and_baru = syarat_nilai and syarat_kehadiran
print(f"Status Kelulusan baru (menggunakan 'and'): {lulus_and_baru}\n")

print("=" * 50 + "\n")

# ==========================================
# Tugas 5: Gunakan operator 'or'
# ==========================================
print("--- PENGGUNAAN OPERATOR 'or' (Nilai: 70, Kehadiran: 90) ---")
lulus_or = syarat_nilai or syarat_kehadiran
print(f"Status Kelulusan (menggunakan 'or'): {lulus_or}")
```
