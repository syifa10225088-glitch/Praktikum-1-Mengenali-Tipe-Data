```python
print("=========================================")
print("  PROGRAM EVALUASI KELULUSAN MAHASISWA")
print("=========================================\n")

# --- PENGUJIAN KELOMPOK 1 (Kombinasi Lulus) ---
# 1. Simpan nama mahasiswa dalam variabel string
nama_1 = "Budi Santoso"

# 2. Simpan nilai tugas dan nilai ujian dalam variabel numerik
nilai_tugas_1 = 80
nilai_ujian_1 = 85

# 3. Hitung rata-rata kedua nilai
rata_rata_1 = (nilai_tugas_1 + nilai_ujian_1) / 2

# 4. Status lulus jika rata-rata minimal 75 (menghasilkan True/False)
status_lulus_1 = rata_rata_1 >= 75

# 5. Tampilkan hasil pengujian 1
print("--- Pengujian 1 ---")
print(f"Nama Mahasiswa : {nama_1}")
print(f"Nilai Tugas    : {nilai_tugas_1}")
print(f"Nilai Ujian    : {nilai_ujian_1}")
print(f"Rata-rata      : {rata_rata_1}")
print(f"Status Lulus   : {status_lulus_1}\n")


# --- PENGUJIAN KELOMPOK 2 (Kombinasi Tidak Lulus) ---
# 1. Simpan nama mahasiswa baru
nama_2 = "Siti Aminah"

# 2. Simpan nilai tugas dan ujian baru
nilai_tugas_2 = 65
nilai_ujian_2 = 70

# 3. Hitung rata-rata kedua nilai
rata_rata_2 = (nilai_tugas_2 + nilai_ujian_2) / 2

# 4. Status lulus jika rata-rata minimal 75
status_lulus_2 = rata_rata_2 >= 75

# 5. Tampilkan hasil pengujian 2
print("--- Pengujian 2 ---")
print(f"Nama Mahasiswa : {nama_2}")
print(f"Nilai Tugas    : {nilai_tugas_2}")
print(f"Nilai Ujian    : {nilai_ujian_2}")
print(f"Rata-rata      : {rata_rata_2}")
print(f"Status Lulus   : {status_lulus_2}")
print("=========================================")
```
