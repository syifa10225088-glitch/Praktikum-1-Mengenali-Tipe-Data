```python
# Langkah 1: Simpan poin awal dalam variabel poin
poin = 100
print(f"1. Nilai awal\t\t: {poin}")

# Langkah 2: Tambahkan 25 poin menggunakan +=
poin += 25
print(f"2. poin += 25\t\t: {poin}")

# Langkah 3: Kurangi 10 poin menggunakan -=
poin -= 10
print(f"3. poin -= 10\t\t: {poin}")

# Langkah 4: Kalikan poin dengan 2 menggunakan *=
poin *= 2
print(f"4. poin *= 2\t\t: {poin}")

# Langkah 5: Bagi poin dengan 2 menggunakan /=
poin /= 2
print(f"5. poin /= 2\t\t: {poin}")
```
