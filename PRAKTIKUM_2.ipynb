```python
# 1. Buat variabel harga dan jumlah
harga = 12500
jumlah = 4

# 2. Hitung total harga
total = harga * jumlah

# 3. Sisa pembagian total harga dengan 2.000
sisa = total % 2000

# 4. Harga setelah diskon Rp5.000
total_diskon = total - 5000

# 5. Tampilkan hasil dengan label yang jelas
print("Harga per buku        : Rp", harga)
print("Jumlah buku           : ", jumlah)
print("Total harga           : Rp", total)
print("Sisa bagi 2.000 (%)   : Rp", sisa)
print("Harga setelah diskon  : Rp", total_diskon)
```
